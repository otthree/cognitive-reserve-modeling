## 4. ADNI Variable Presence Check

ADNI merged dataset에 있던 6개 변수가 HCP aging에도 있는지 확인.

| ADNI 변수 | 유형 | 설명 |
|-----------|------|------|
| FDG_BL | 연속형 스칼라 | 고정밀 바이오마커 |
| ADASQ4 | 순서형 점수 | 인지 기능 직접 반영 |
| VSBPDIA | 연속형 스칼라 | 혈압 (약한 AD 신호) |
| VSTEM | 연속형 스칼라 | 체온 (약한 AD 신호) |
| VSRESP | 범주형/플래그 | 데이터 가용성 지시자 |
| ADNI_VS | 범주형/플래그 | 데이터 가용성 지시자 |

```python
adni_vars = ['FDG_BL', 'ADASQ4', 'VSBPDIA', 'VSTEM', 'VSRESP', 'ADNI_VS']

all_short_lower = {c.lower(): c for c in df_short.columns}

print("=== ADNI variables in HCP aging ===")
for var in adni_vars:
    exact = var.lower() in all_short_lower
    partial = [c for c in df.columns if var.lower() in c.lower()]
    status = "FOUND" if exact else ("PARTIAL MATCH" if partial else "NOT FOUND")
    print(f"  {var:12s}: {status}", end="")
    if partial:
        print(f"  -> {partial[:3]}")
    else:
        print()
```
